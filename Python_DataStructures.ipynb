## 1️⃣ Lists
*(see Definitions.md → Lists for the why/when/function table)*

```python
# ------------------------------------------------------------- #
# 🔹 1. LISTS — DEMO
# ------------------------------------------------------------- #

students = ["Anika", "Bhavya", "Chetan", "Divya"]

students.append("Esha")
print("After append:", students)

students.insert(1, "Farah")
print("After insert:", students)

students.remove("Chetan")
print("After remove:", students)

last_student = students.pop()
print("Removed via pop:", last_student, "| Remaining:", students)

students.sort()
print("Sorted:", students)

print("Total students:", len(students))

top_two = students[:2]
print("Top two (sliced):", top_two)
```

## 2️⃣ Tuples
*(see Definitions.md → Tuples for the why/when/function table)*

```python
# ------------------------------------------------------------- #
# 🔹 2. TUPLES — DEMO
# ------------------------------------------------------------- #

student_record = (101, "Anika", "CSE")

roll_no, name, dept = student_record
print(f"Roll: {roll_no}, Name: {name}, Dept: {dept}")

try:
    student_record[1] = "Bhavya"
except TypeError as e:
    print("Cannot modify a tuple! Error:", e)

def get_min_max(numbers):
    return min(numbers), max(numbers)

lowest, highest = get_min_max([56, 78, 90, 45])
print("Lowest:", lowest, "| Highest:", highest)

grades = ("A", "B", "A", "C", "A", "B")
print("Number of A grades:", grades.count("A"))
print("Position of first C grade:", grades.index("C"))

seat_map = {(1, "A"): "Anika", (1, "B"): "Farah"}
print("Seat 1A belongs to:", seat_map[(1, "A")])
```

## 3️⃣ Dictionaries
*(see Definitions.md → Dictionaries for the why/when/function table)*

```python
# ------------------------------------------------------------- #
# 🔹 3. DICTIONARIES — DEMO
# ------------------------------------------------------------- #

marks = {"Anika": 88, "Bhavya": 92, "Chetan": 76}

marks["Divya"] = 95
marks["Chetan"] = 80

score = marks.get("Ravi", "Not graded yet")
print("Ravi's score:", score)

if "Anika" in marks:
    print("Anika's mark is on record:", marks["Anika"])

for name, mark in marks.items():
    print(f"{name} scored {mark}")

average = sum(marks.values()) / len(marks)
print("Class average:", round(average, 2))

marks.pop("Bhavya")
print("After withdrawal:", marks)

new_batch = {"Esha": 89, "Farah": 91}
marks.update(new_batch)
print("After merging new batch:", marks)
```

## 4️⃣ Sets
*(see Definitions.md → Sets for the why/when/function table and complexity note)*

```python
# ------------------------------------------------------------- #
# 🔹 4. SETS — DEMO
# ------------------------------------------------------------- #

workshop_a = {"Anika", "Bhavya", "Chetan", "Divya"}
workshop_b = {"Chetan", "Divya", "Esha", "Farah"}

workshop_a.add("Anika")
print("Workshop A (no duplicate added):", workshop_a)

all_students = workshop_a | workshop_b
print("All students (union):", all_students)

both_workshops = workshop_a & workshop_b
print("Students in both (intersection):", both_workshops)

only_a = workshop_a - workshop_b
print("Only in Workshop A (difference):", only_a)

print("Is Farah in Workshop A?", "Farah" in workshop_a)

workshop_a.discard("Bhavya")
print("After discard:", workshop_a)
```

### 📝 Practice — Sets

## 5️⃣ File Handling
*(see Definitions.md → File Handling for the why/when/function table)*

```python
# ------------------------------------------------------------- #
# 🔹 5. FILE HANDLING — DEMO
# ------------------------------------------------------------- #

with open("student_report.txt", "w") as f:
    f.write("Bootcamp Student Report\n")
    f.write("========================\n")
    lines = ["Anika: 88\n", "Bhavya: 92\n", "Chetan: 76\n"]
    f.writelines(lines)

with open("student_report.txt", "a") as f:
    f.write("Divya: 95 (late entry)\n")

with open("student_report.txt", "r") as f:
    full_content = f.read()
print("----- Full file content -----")
print(full_content)

with open("student_report.txt", "r") as f:
    all_lines = f.readlines()
print("Number of lines in file:", len(all_lines))

with open("student_report.txt", "r") as f:
    first_line = f.readline()
print("First line only:", first_line.strip())
```

```python
# ------------------------------------------------------------- #
# 🔹 5. FILE HANDLING — MISSING FILE (try/except)
# ------------------------------------------------------------- #

try:
    with open("does_not_exist.txt", "r") as f:
        content = f.read()
except FileNotFoundError:
    print("That file doesn't exist yet -- creating it now with default content.")
    with open("does_not_exist.txt", "w") as f:
        f.write("This file was just created.\n")

with open("does_not_exist.txt", "r") as f:
    print("Content now:", f.read().strip())
```
