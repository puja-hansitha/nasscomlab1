### B3. Conditionals — Discount + Parcel Charge
Add a loyalty discount: >Rs.200 gets 10% off, >Rs.100 gets 5% off, else no discount. Add a Rs.5 packing charge if `is_parcel = True`. Test with at least 3 different total values.

```python
# B3 solution
def calculate_final_bill(total, is_parcel):
    if total > 200:
        discount = total * 0.10
    elif total > 100:
        discount = total * 0.05
    else:
        discount = 0
    packing_charge = 5 if is_parcel else 0
    final_total = total - discount + packing_charge
    print(f"Total: {total} | Discount: {discount} | Packing: {packing_charge} | Final: {final_total}")

calculate_final_bill(250, True)   # > 200 band
calculate_final_bill(150, False)  # > 100 band
calculate_final_bill(80, True)    # no discount
```

### B8. Functions — `calculate_bill()`
Write `calculate_bill(cart, menu, is_parcel=False, gst_rate=0.05)` that applies the discount logic from B3 and returns a dict with `subtotal`, `discount`, `gst`, `packing_charge`, and `total`. Call it for two different carts.

```python
# B8 solution
def calculate_bill(cart, menu, is_parcel=False, gst_rate=0.05):
    subtotal = sum(menu[item] for item in cart)
    if subtotal > 200:
        discount = subtotal * 0.10
    elif subtotal > 100:
        discount = subtotal * 0.05
    else:
        discount = 0
    gst = (subtotal - discount) * gst_rate
    packing_charge = 5 if is_parcel else 0
    total = subtotal - discount + gst + packing_charge
    return {"subtotal": subtotal, "discount": discount, "gst": round(gst, 2),
            "packing_charge": packing_charge, "total": round(total, 2)}

full_menu = {"Samosa": 15.0, "Vada": 12.0, "Tea": 10.0,
             "Cold Coffee": 30.0, "Dosa": 40.0, "Idli": 20.0, "Thali": 150.0}

print(calculate_bill(["Samosa", "Dosa", "Cold Coffee"], full_menu, is_parcel=True))
print(calculate_bill(["Thali", "Thali"], full_menu))
```

### B9. File Handling — Daily Sales Log
Write `save_receipt(customer_name, bill)` that appends `"<name>: Rs.<total>"` to `my_canteen_sales_log.txt`. Call it for at least two orders, then read and print the full log.

```python
# B9 solution
def save_receipt(customer_name, bill):
    with open("my_canteen_sales_log.txt", "a") as f:
        f.write(f"{customer_name}: Rs.{bill['total']:.2f}\n")

bill_ravi   = calculate_bill(["Samosa", "Tea"], full_menu)
bill_ananya = calculate_bill(["Thali", "Thali"], full_menu)

save_receipt("Ravi", bill_ravi)
save_receipt("Ananya", bill_ananya)

print("--- Sales Log ---")
with open("my_canteen_sales_log.txt") as f:
    print(f.read())
```

### B10. Exception Handling — Safer Billing Counter
Write `bill_and_collect_cash(cart, menu, cash_paid, is_parcel=False)` that: (a) catches `KeyError` for unknown items, (b) catches `ValueError` for insufficient cash, (c) prints change due on success. Test all three cases.

```python
# B10 solution
def bill_and_collect_cash(cart, menu, cash_paid, is_parcel=False):
    try:
        for item in cart:
            if item not in menu:
                raise KeyError(f"'{item}' is not on the menu")
        bill = calculate_bill(cart, menu, is_parcel)
        if cash_paid < bill["total"]:
            raise ValueError(f"Need Rs.{bill['total']:.2f}, got Rs.{cash_paid}")
        print(f"Change due: Rs.{cash_paid - bill['total']:.2f}")
    except KeyError as e:
        print(f"KeyError: {e}")
    except ValueError as e:
        print(f"ValueError: {e}")

print("--- Valid order ---")
bill_and_collect_cash(["Samosa", "Tea"], full_menu, 50)

print("--- Invalid item ---")
bill_and_collect_cash(["Samosa", "Pizza"], full_menu, 100)

print("--- Insufficient cash ---")
bill_and_collect_cash(["Dosa", "Vada"], full_menu, 20)
```

## Part C — End-of-Lab Combined Task

Bring B1–B10 together into one working canteen billing program that:
1. Defines a **menu dictionary** with at least 6 items.
2. Builds a **cart** for each customer.
3. Uses `calculate_bill()` with discount and packing charge logic.
4. **Saves** each receipt to `my_canteen_sales_log.txt`.
5. Wraps each transaction in **try/except**.
6. Bills **at least 3 different customers** in a row, then prints the full day's log.

```python
# Part C solution
import os

# Reset log
if os.path.exists("my_canteen_sales_log.txt"):
    os.remove("my_canteen_sales_log.txt")

menu = {
    "Samosa": 15.0, "Vada": 12.0, "Tea": 10.0,
    "Cold Coffee": 30.0, "Dosa": 40.0, "Idli": 20.0, "Thali": 150.0
}

def process_customer(name, cart, cash_paid, is_parcel=False):
    print(f"--- {name} ---")
    try:
        for item in cart:
            if item not in menu:
                raise KeyError(f"'{item}' not on menu")
        bill = calculate_bill(cart, menu, is_parcel)
        if cash_paid < bill["total"]:
            raise ValueError(f"Insufficient cash (need Rs.{bill['total']:.2f})")
        save_receipt(name, bill)
        print(f"Total: Rs.{bill['total']:.2f} | Change: Rs.{cash_paid - bill['total']:.2f}")
    except (KeyError, ValueError) as e:
        print(f"Order failed: {e}")

process_customer("Ravi",   ["Samosa", "Tea"],          cash_paid=50)
process_customer("Ananya", ["Dosa", "Cold Coffee"],    cash_paid=200, is_parcel=True)
process_customer("Kiran",  ["Thali", "Thali"],         cash_paid=300)
process_customer("Meena",  ["Idli", "Pizza"],          cash_paid=100)  # bad item

print("\n--- Full Day's Log ---")
with open("my_canteen_sales_log.txt") as f:
    print(f.read())
```

## Stretch Challenge — Canteen Rush Hour System

### Requirements
Write `process_order(order, menu)` and a driver loop that:
1. **Peak-hour surcharge** — hour 12 or 13: +8% on subtotal before GST.
2. **Two competing discounts** — membership (`regular`: 0%, `silver`: 5%, `gold`: 12%) vs loyalty (>200→10%, >100→5%). Apply the **larger** one only.
3. **GST 5%** after discount + Rs.5 packing charge if `is_parcel`.
4. **Bad orders** → log to `failed_transactions.txt`, continue loop.
5. **Successful orders** → append to `rush_hour_log.txt`.
6. **End-of-rush summary**: total revenue, revenue by category, best-selling item, failed order count.

```python
# Stretch Challenge solution
import os

menu_rich = {
    "Samosa":     {"price": 15.0, "category": "Snacks"},
    "Vada":       {"price": 12.0, "category": "Snacks"},
    "Tea":        {"price": 10.0, "category": "Beverages"},
    "Cold Coffee":{"price": 30.0, "category": "Beverages"},
    "Dosa":       {"price": 40.0, "category": "Meals"},
    "Idli":       {"price": 20.0, "category": "Meals"},
}

orders = [
    {"customer": "Ravi",   "cart": ["Samosa", "Tea"],        "tier": "regular", "hour": 13, "is_parcel": False, "cash_paid": 30},
    {"customer": "Ananya", "cart": ["Dosa", "Cold Coffee"],  "tier": "gold",    "hour": 12, "is_parcel": True,  "cash_paid": 100},
    {"customer": "Kiran",  "cart": ["Vada", "Idli"],         "tier": "silver",  "hour": 14, "is_parcel": False, "cash_paid": 45},
    {"customer": "Meena",  "cart": ["Pizza"],                "tier": "regular", "hour": 13, "is_parcel": False, "cash_paid": 100},
    {"customer": "Arjun",  "cart": ["Dosa", "Dosa", "Tea"],  "tier": "gold",    "hour": 12, "is_parcel": False, "cash_paid": 95},
    {"customer": "Divya",  "cart": ["Tea"],                  "tier": "regular", "hour": 15, "is_parcel": False, "cash_paid": 5},
]

for f in ["rush_hour_log.txt", "failed_transactions.txt"]:
    if os.path.exists(f): os.remove(f)

total_revenue = 0
revenue_by_category = {"Snacks": 0.0, "Beverages": 0.0, "Meals": 0.0}
item_sales = {}
failed_orders = []

MEMBERSHIP_RATES = {"regular": 0.0, "silver": 0.05, "gold": 0.12}

def process_order(order, menu):
    global total_revenue
    try:
        for item in order["cart"]:
            if item not in menu:
                raise KeyError(f"'{item}' not on menu")

        subtotal = sum(menu[i]["price"] for i in order["cart"])
        surcharge = subtotal * 0.08 if order["hour"] in (12, 13) else 0
        base = subtotal + surcharge

        loyalty = 0.10 if base > 200 else (0.05 if base > 100 else 0)
        discount = base * max(MEMBERSHIP_RATES[order["tier"]], loyalty)

        gst = (base - discount) * 0.05
        packing = 5 if order["is_parcel"] else 0
        total = base - discount + gst + packing

        if order["cash_paid"] < total:
            raise ValueError(f"Insufficient cash (need Rs.{total:.2f})")

        # Log success
        with open("rush_hour_log.txt", "a") as f:
            f.write(f"{order['customer']}: Rs.{total:.2f}\n")

        # Update stats
        total_revenue += total
        for item in order["cart"]:
            revenue_by_category[menu[item]["category"]] += menu[item]["price"]
            item_sales[item] = item_sales.get(item, 0) + 1

    except (KeyError, ValueError) as e:
        reason = str(e)
        failed_orders.append({"customer": order["customer"], "reason": reason})
        with open("failed_transactions.txt", "a") as f:
            f.write(f"{order['customer']}: {reason}\n")

for order in orders:
    process_order(order, menu_rich)

# Summary
print("=== End-of-Rush Summary ===")
print(f"Total Revenue: Rs.{total_revenue:.2f}")
print("\nRevenue by Category:")
for cat, amt in revenue_by_category.items():
    print(f"  {cat}: Rs.{amt:.2f}")
best = max(item_sales, key=item_sales.get)
print(f"\nBest-Selling Item: {best} (x{item_sales[best]})")
print(f"\nFailed Orders: {len(failed_orders)}")
for fo in failed_orders:
    print(f"  {fo['customer']}: {fo['reason']}")
```
