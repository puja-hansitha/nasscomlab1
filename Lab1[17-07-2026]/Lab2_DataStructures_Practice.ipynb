### 1. Lists — Lab Submissions Log

### 5. Sets — Comparing Two Students' Skills
Create `your_skills` and `friend_skills` sets (at least 4 items each, with some overlap). Print intersection, union, and difference.

```python
# 5. Sets solution
your_skills    = {"Python", "Java", "C++", "Git"}
friend_skills  = {"Python", "JavaScript", "HTML", "Git"}

print("Both know:", your_skills & friend_skills)
print("All skills:", your_skills | friend_skills)
print("Only I know:", your_skills - friend_skills)
```

### 6. Dictionaries — Nested Gradebook
Create a nested dict `gradebook` (at least 5 subjects, each with `marks`, `credits`, `faculty`). Print one subject's marks via double-key lookup. Use `.get()` for a missing subject.

### 7. Dictionary Comprehension
Using `gradebook` from Exercise 6, build a `{subject: marks}` dict in one line using a dictionary comprehension.

```python
# 7. Dictionary comprehension solution
gradebook = {
    "OS":   {"marks": 85, "credits": 3, "faculty": "Dr. A"},
    "DBMS": {"marks": 92, "credits": 3, "faculty": "Dr. B"},
    "CN":   {"marks": 78, "credits": 4, "faculty": "Dr. C"},
    "DAA":  {"marks": 88, "credits": 4, "faculty": "Dr. D"},
    "SE":   {"marks": 95, "credits": 3, "faculty": "Dr. E"},
}

subject_marks = {sub: details["marks"] for sub, details in gradebook.items()}
print(subject_marks)
```

### 8. Set Comprehension — High-Scoring Subjects
Using `gradebook` (Ex 6) and unique subjects submitted (Ex 4), build a set comprehension keeping only submitted subjects with marks >= 85.

```python
# 8. Set comprehension solution
gradebook = {
    "OS":   {"marks": 85, "credits": 3, "faculty": "Dr. A"},
    "DBMS": {"marks": 92, "credits": 3, "faculty": "Dr. B"},
    "CN":   {"marks": 78, "credits": 4, "faculty": "Dr. C"},
    "DAA":  {"marks": 88, "credits": 4, "faculty": "Dr. D"},
    "SE":   {"marks": 95, "credits": 3, "faculty": "Dr. E"},
}
unique_subjects_submitted = {"OS", "DBMS", "CN", "DAA", "SE"}

high_scoring = {
    sub for sub, d in gradebook.items()
    if sub in unique_subjects_submitted and d["marks"] >= 85
}
print(high_scoring)
```

## Combined Task — Semester Report (submit this)

Bring everything together into one program that:
1. Defines a nested `gradebook` dict (at least 6 subjects, with `marks` and `credits`).
2. Defines `lab_submissions` list with at least 10 entries (with repeats).
3. Computes **unique subjects submitted** using a set.
4. Computes **high-scoring subjects (marks >= 85)** using a set comprehension.
5. Computes **best subject** as a tuple `(subject_name, marks)`.
6. Prints all of the above in a clearly labelled semester report.

```python
# Combined Task solution

# 1. Gradebook
gradebook = {
    "OS":   {"marks": 85, "credits": 3},
    "DBMS": {"marks": 92, "credits": 3},
    "CN":   {"marks": 78, "credits": 4},
    "DAA":  {"marks": 88, "credits": 4},
    "SE":   {"marks": 95, "credits": 3},
    "AI":   {"marks": 91, "credits": 3},
}

# 2. Lab submissions
lab_submissions = ["OS", "DBMS", "CN", "OS", "DAA", "DBMS", "OS", "SE", "AI", "AI"]

# 3. Unique subjects submitted
unique_submitted = set(lab_submissions)

# 4. High-scoring subjects
high_scoring = {sub for sub, d in gradebook.items() if d["marks"] >= 85}

# 5. Best subject tuple
best_subject = max(gradebook.items(), key=lambda x: x[1]["marks"])
best_tuple = (best_subject[0], best_subject[1]["marks"])

# 6. Print report
print("=== Semester Report ===")
print(f"Unique Subjects Submitted ({len(unique_submitted)}): {unique_submitted}")
print(f"High-Scoring Subjects (>=85): {high_scoring}")
print(f"Best Subject: {best_tuple[0]} with {best_tuple[1]} marks")
```
