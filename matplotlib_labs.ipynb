## Section 1 — Line Plots: `y = f(x)`

**Math idea:** a line plot samples a function `y = f(x)` at discrete `x` values and connects the dots — the more points, the smoother the curve looks.

````python
# WORKED EXAMPLE
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

plt.plot(x, y)
plt.xlabel("x (radians)")
plt.ylabel("sin(x)")
plt.title("y = sin(x)")
plt.show()
````

### Try it yourself
1. Plot `y = cos(x)` over the same range as the worked example.
2. Add `plt.xlabel`, `plt.ylabel`, and `plt.title` to label your plot.
3. Plot both `sin(x)` and `cos(x)` on the *same* axes (call `plt.plot` twice before `plt.show()`).

````python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0, 2 * np.pi, 100)
y_sin = np.sin(x)
y_cos = np.cos(x)

# Plot y = cos(x)
plt.plot(x, y_cos, label="cos(x)")
plt.xlabel("x (radians)")
plt.ylabel("cos(x)")
plt.title("y = cos(x)")
plt.legend()
plt.show()

# Plot both sin(x) and cos(x) on the same axes
plt.plot(x, y_sin, label="sin(x)")
plt.plot(x, y_cos, label="cos(x)")
plt.xlabel("x (radians)")
plt.ylabel("y")
plt.title("y = sin(x) and y = cos(x)")
plt.legend()
plt.show()
````

## Section 2 — Scatter Plots: Points in a Vector Space

**Math idea:** a scatter plot places every `(x, y)` pair as a literal point in 2-D space — the same geometric picture as a NumPy vector `[x, y]`.

````python
# WORKED EXAMPLE
rng = np.random.default_rng(0)
x = rng.normal(0, 1, 50)
y = 2 * x + rng.normal(0, 0.5, 50)

plt.scatter(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Correlated data")
plt.show()
````

### Try it yourself
1. Generate 50 random points where `y` is *unrelated* to `x` (e.g. `y = rng.normal(0, 1, 50)` independently).
2. Scatter-plot them and compare visually to the worked example — which one looks correlated?
3. Compute `np.corrcoef(x, y)` for both cases to check your eyes were right.

````python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(0)

# Worked example's correlated data for comparison
x_corr = rng.normal(0, 1, 50)
y_corr = 2 * x_corr + rng.normal(0, 0.5, 50)

# 1. Generate 50 random points where y is unrelated to x
x_unrelated = rng.normal(0, 1, 50)
y_unrelated = rng.normal(0, 1, 50)

# 2. Scatter-plot them and compare visually
plt.scatter(x_unrelated, y_unrelated)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Uncorrelated data")
plt.show()

# 3. Compute np.corrcoef(x, y) for both cases
print("Correlation coefficient for correlated data (worked example):\n", np.corrcoef(x_corr, y_corr))
print("\nCorrelation coefficient for uncorrelated data (try it yourself):\n", np.corrcoef(x_unrelated, y_unrelated))
````

## Section 3 — Histograms: Discretized Distributions

**Math idea:** a histogram bins continuous data and counts values per bin — a discrete approximation of a probability density function.

````python
# WORKED EXAMPLE
data = rng.normal(loc=50, scale=10, size=1000)

plt.hist(data, bins=30, edgecolor="black")
plt.xlabel("value")
plt.ylabel("count")
plt.title("Distribution of data (~Normal(50, 10))")
plt.show()

print("mean:", data.mean(), "std:", data.std())
````

### Try it yourself
1. Generate 1000 samples from a *uniform* distribution: `rng.uniform(0, 100, 1000)`.
2. Plot a histogram with 30 bins and compare its shape to the normal distribution above.
3. Try `bins=5` and `bins=100` on the same data — how does bin count change what you can see?

````python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(0)

# 1. Generate 1000 samples from a uniform distribution
data_uniform = rng.uniform(0, 100, 1000)

# 2. Plot a histogram with 30 bins
plt.hist(data_uniform, bins=30, edgecolor="black")
plt.xlabel("value")
plt.ylabel("count")
plt.title("Distribution of data (~Uniform(0, 100)) - 30 bins")
plt.show()

# 3. Try bins=5 and bins=100 on the same data
plt.hist(data_uniform, bins=5, edgecolor="black")
plt.xlabel("value")
plt.ylabel("count")
plt.title("Distribution of data (~Uniform(0, 100)) - 5 bins")
plt.show()

plt.hist(data_uniform, bins=100, edgecolor="black")
plt.xlabel("value")
plt.ylabel("count")
plt.title("Distribution of data (~Uniform(0, 100)) - 100 bins")
plt.show()
````

## Section 4 — Bar Charts: Categorical Aggregation

**Math idea:** a bar chart visualizes a function from a finite set of categories to a number — exactly the shape of a `groupby(...).agg(...)` result.

````python
# WORKED EXAMPLE
subjects = ["Math", "Science", "English"]
avg_scores = [78.4, 82.1, 75.6]

plt.bar(subjects, avg_scores, color=["#4C72B0", "#55A868", "#C44E52"])
plt.ylabel("Average score")
plt.title("Average score per subject")
plt.show()
````

## Section 5 — Box Plots: The Five-Number Summary

**Math idea:** a box plot draws the five-number summary (min, Q1, median, Q3, max) plus outliers in one compact shape, making it easy to compare spread across groups.

````python
# WORKED EXAMPLE
groups = [rng.normal(0, 1, 100), rng.normal(2, 1.5, 100), rng.normal(-1, 0.5, 100)]

plt.boxplot(groups, tick_labels=["A", "B", "C"])
plt.ylabel("value")
plt.title("Spread comparison across groups")
plt.show()
````

### Try it yourself
1. Create three groups of exam scores (as lists or arrays) with different means and spreads.
2. Draw a box plot comparing all three.
3. Which group has the widest interquartile range (the box itself, not the whiskers)?

````python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(0)

# 1. Create three groups of exam scores with different means and spreads.
group1 = rng.normal(70, 5, 50)  # Mean 70, Std Dev 5
group2 = rng.normal(80, 10, 50) # Mean 80, Std Dev 10
group3 = rng.normal(60, 3, 50)  # Mean 60, Std Dev 3

groups_data = [group1, group2, group3]

# 2. Draw a box plot comparing all three.
plt.boxplot(groups_data, tick_labels=["Group 1", "Group 2", "Group 3"])
plt.ylabel("Exam Score")
plt.title("Exam Score Distribution Across Three Groups")
plt.show()

# 3. Which group has the widest interquartile range?
# Calculate Q1 and Q3 for each group and then the IQR
iqrs = []
for i, group in enumerate(groups_data):
    q1 = np.percentile(group, 25)
    q3 = np.percentile(group, 75)
    iqr = q3 - q1
    iqrs.append(iqr)
    print(f"Group {i+1} IQR: {iqr:.2f}")

widest_iqr_group = np.argmax(iqrs) + 1
print(f"\nGroup {widest_iqr_group} has the widest interquartile range.")
````

## Section 6 — Heatmaps: Visualizing a Matrix

**Math idea:** `imshow` renders a 2-D array directly as a grid of colored cells — the most direct possible visualization of a matrix, such as a correlation matrix.

````python
# WORKED EXAMPLE
matrix = np.array([
    [1.0, 0.8, 0.2],
    [0.8, 1.0, 0.5],
    [0.2, 0.5, 1.0],
])

plt.imshow(matrix, cmap="viridis", vmin=-1, vmax=1)
plt.colorbar(label="correlation")
plt.xticks(range(3), ["Math", "Science", "English"])
plt.yticks(range(3), ["Math", "Science", "English"])
plt.title("Correlation matrix")
plt.show()
````

2. Display it with `plt.imshow` using the `"plasma"` colormap.

````python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(0)

# 1. Build a 4x4 matrix of random values between 0 and 1
random_matrix = rng.uniform(0, 1, (4, 4))

# 2. Display it with plt.imshow using the "plasma" colormap.
plt.imshow(random_matrix, cmap="plasma", vmin=0, vmax=1)

# 3. Add a plt.colorbar()
plt.colorbar(label="Random Value")

# Optional: Add tick labels for clarity
plt.xticks(range(4), [f"Col {i+1}" for i in range(4)])
plt.yticks(range(4), [f"Row {i+1}" for i in range(4)])

plt.title("4x4 Matrix of Random Values")
plt.show()
````

## Section 7 — Subplots: Multiple Coordinate Planes in One Figure

**Math idea:** subplots put several independent coordinate planes on one canvas, useful for direct side-by-side comparison.

````python
# WORKED EXAMPLE
x = np.linspace(0, 2 * np.pi, 100)
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].plot(x, np.sin(x))
axes[0].set_title("sin(x)")

axes[1].plot(x, np.cos(x))
axes[1].set_title("cos(x)")

plt.tight_layout()
plt.show()
````

### Try it yourself
1. Create a figure with a **2x2** grid of subplots (`plt.subplots(2, 2, figsize=(8, 6))`).
2. Plot a different chart type (line, scatter, hist, bar) in each of the four axes.
3. Give each subplot its own `set_title`, then call `plt.tight_layout()` before `plt.show()`.

````python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(0)

# Data for subplots
x = np.linspace(0, 2 * np.pi, 50)
y_sin = np.sin(x)
y_cos = np.cos(x)

scatter_x = rng.normal(0, 1, 30)
scatter_y = rng.normal(0, 1, 30)

hist_data = rng.exponential(1, 1000)

bar_categories = ['A', 'B', 'C']
bar_values = [10, 25, 15]

# 1. Create a figure with a 2x2 grid of subplots
fig, axes = plt.subplots(2, 2, figsize=(8, 6))

# 2. Plot a different chart type in each of the four axes.
# Top-left: Line plot
axes[0, 0].plot(x, y_sin)
axes[0, 0].set_title("Line Plot: sin(x)")

# Top-right: Scatter plot
axes[0, 1].scatter(scatter_x, scatter_y)
axes[0, 1].set_title("Scatter Plot: Random Data")

# Bottom-left: Histogram
axes[1, 0].hist(hist_data, bins=20, edgecolor='black')
axes[1, 0].set_title("Histogram: Exponential Data")

# Bottom-right: Bar chart
axes[1, 1].bar(bar_categories, bar_values, color=['skyblue', 'lightcoral', 'lightgreen'])
axes[1, 1].set_title("Bar Chart: Categorical Values")

# 3. Call plt.tight_layout() before plt.show()
plt.tight_layout()
plt.show()
````

## Mini-Project — Put it all together

**Scenario:** You have exam scores for 5 students across 3 subjects (Math, Science, English) — the same dataset from the NumPy/Pandas mini-projects. Visualize it fully using only Matplotlib.

```python
subjects = ["Math", "Science", "English"]
scores = np.array([
    [85, 90, 78],
    [70, 65, 80],
    [90, 88, 95],
    [60, 72, 68],
    [75, 80, 85],
])  # rows = students, columns = subjects
```

Tasks:
1. Bar chart of each subject's **average** score (column means) — one bar per subject.
2. Box plot comparing the **distribution** of scores per subject (pass `scores` directly to `plt.boxplot`, it accepts a 2-D array column-wise).
3. Heatmap of the **correlation matrix** between subjects (`np.corrcoef(scores, rowvar=False)`), with a colorbar and axis tick labels.
4. Combine all three into one figure using `plt.subplots(1, 3, figsize=(15, 4))`, one chart per axis.

````python
import numpy as np
import matplotlib.pyplot as plt
subjects = ["Math", "Science", "English"]
scores = np.array([
    [85, 90, 78],
    [70, 65, 80],
    [90, 88, 95],
    [60, 72, 68],
    [75, 80, 85],
])  # rows = students, columns = subjects

# 4. Combine all three into one figure using plt.subplots(1, 3, figsize=(15, 4))
fig, axes = plt.subplots(1, 3, figsize=(18, 5)) # Adjusted figsize for better visibility

# Task 1: Bar chart of each subject's average score (column means)
avg_scores = np.mean(scores, axis=0)
axes[0].bar(subjects, avg_scores, color=["#4C72B0", "#55A868", "#C44E52"])
axes[0].set_ylabel("Average Score")
axes[0].set_title("Average Score per Subject")
axes[0].set_ylim(0, 100) # Ensure y-axis starts at 0 for comparison

# Task 2: Box plot comparing the distribution of scores per subject
axes[1].boxplot(scores, tick_labels=subjects)
axes[1].set_ylabel("Score Distribution")
axes[1].set_title("Score Distribution per Subject (Box Plot)")
axes[1].set_ylim(0, 100) # Ensure y-axis starts at 0

# Task 3: Heatmap of the correlation matrix between subjects
correlation_matrix = np.corrcoef(scores, rowvar=False)
im = axes[2].imshow(correlation_matrix, cmap="viridis", vmin=-1, vmax=1)
axes[2].set_xticks(range(len(subjects)), labels=subjects, rotation=45, ha='right')
axes[2].set_yticks(range(len(subjects)), labels=subjects)
axes[2].set_title("Correlation Matrix between Subjects")
fig.colorbar(im, ax=axes[2], label="Correlation")

plt.tight_layout()
plt.show()
````
